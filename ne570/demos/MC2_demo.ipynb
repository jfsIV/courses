```python
import matplotlib.pyplot as plt
import numpy as np
from mpl_toolkits.mplot3d import Axes3D

plt.rcParams.update({'font.size': 25})
```

```python
def lcg(x_now, a, b, M):
    return (a*x_now + b) % M
```

# Example: a =5, b=1, M=16, x_0=1

```python
a   = 5
b   = 1
M   = np.power(2,4)
x_0 = 1

PRNG_seq_change_a_1 = np.zeros(M+1)
PRNG_seq_change_a_1[0] = x_0

for i in range(1, M+1):
    PRNG_seq_change_a_1[i] = lcg(PRNG_seq_change_a_1[i-1], a, b, M)
```

```python
range_plot = np.arange(len(PRNG_seq_change_a_1))
spread_plot = np.linspace(0, 2*np.pi, len(PRNG_seq_change_a_1)+1)

fig = plt.figure(figsize=(12,12))
ax = fig.add_subplot(1, 1, 1, polar=True)
plt.thetagrids(np.degrees(spread_plot[0:-1]), range_plot)
ax.plot(spread_plot[0:-1], PRNG_seq_change_a_1, 'k-', marker='o', mfc='r', mec='r', ms=8)
ax.set(theta_direction=-1, theta_zero_location='N')

#plt.savefig("PRNG_LCG_a5_b1_M16_x0_1.pdf", bbox_inches='tight')
plt.show()
```
